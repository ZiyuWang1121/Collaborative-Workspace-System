```python
from graphviz import Digraph

def create_vertical_system_diagram():
    """
    Generates a vertical, stacked System Function Call Diagram, restoring all original nodes.
    Target aspect ratio: Tall and Narrow.
    """
    dot = Digraph(
        comment='Project Management System - Vertical Stack',
        format='png'
    )

    # 1. Layout Settings for Verticality
    dot.attr(
        rankdir='TB',        # Top to Bottom (Vertical Layout)
        splines='ortho',     # Orthogonal lines (cleaner corners)
        nodesep='0.6',       # Space between nodes
        ranksep='0.8',       # Space between layers
        concentrate='true',  # Merge lines to reduce clutter
        dpi='300'            # High resolution
    )

    # 2. Node Styles
    dot.attr('node', shape='box', style='rounded,filled', fontname='Arial', fontsize='10', margin='0.2')
    dot.attr('edge', fontname='Arial', fontsize='9')

    # ======================================================
    # A. CORE SYSTEM (Top Level)
    # ======================================================
    with dot.subgraph(name='cluster_0_core') as c:
        c.attr(label='1. Core System', style='rounded', bgcolor='#E3F2FD')
        c.node('PMS_Init', 'ProjectManagementSystem.__init__()', fillcolor='#2196F3', fontcolor='white')
        c.node('PMS_Run', 'ProjectManagementSystem.run()', fillcolor='#2196F3', fontcolor='white')
        c.edge('PMS_Init', 'PMS_Run', label='starts')

    # ======================================================
    # B. UI LAYER (Second Level - All main render functions)
    # ======================================================
    with dot.subgraph(name='cluster_1_ui') as c:
        c.attr(label='2. User Interface', style='rounded', bgcolor='#E1F5FE')

        c.node('Dashboard_UI', 'render_dashboard()', fillcolor='#B3E5FC')
        c.node('Emp_UI', 'render_employee_management()', fillcolor='#B3E5FC')
        c.node('Proj_UI', 'render_project_workflow()', fillcolor='#B3E5FC')
        c.node('Task_UI', 'render_task_workflow()', fillcolor='#B3E5FC')
        c.node('Analytics_UI', 'render_analytics_reports()', fillcolor='#B3E5FC') # Corrected name
        c.node('Docs_UI', 'render_documents_issues()', fillcolor='#B3E5FC')     # Re-added
        c.node('Sidebar_UI', 'render_sidebar()', fillcolor='#B3E5FC')           # Added sidebar render

    # ======================================================
    # C. WORKFLOWS (Third Level)
    # ======================================================
    with dot.subgraph(name='cluster_2_workflows') as c:
        c.attr(label='3. Workflows', style='rounded', bgcolor='#FFFDE7')

        # Project Workflow
        c.node('Proj_Select', 'Proj: Select Action', fillcolor='#FFF9C4')
        c.node('Proj_Edit', 'Proj: Edit Details', fillcolor='#FFF9C4')
        c.node('Proj_Rec', 'Proj: AI Recommendation', fillcolor='#FFF9C4')
        c.node('Proj_Assign', 'Proj: Assignment', fillcolor='#FFF9C4')
        c.edge('Proj_Select', 'Proj_Edit')
        c.edge('Proj_Edit', 'Proj_Rec')
        c.edge('Proj_Rec', 'Proj_Assign')

        # Task Workflow
        c.node('Task_Select', 'Task: Select Action', fillcolor='#FFF9C4')
        c.node('Task_Details', 'Task: Edit Details', fillcolor='#FFF9C4')
        c.node('Task_Predict', 'Task: AI Prediction', fillcolor='#FFF9C4')
        c.node('Task_Assign', 'Task: Assignment', fillcolor='#FFF9C4')
        c.edge('Task_Select', 'Task_Details')
        c.edge('Task_Details', 'Task_Predict')
        c.edge('Task_Predict', 'Task_Assign')

    # ======================================================
    # D. BUSINESS LOGIC (Fourth Level)
    # ======================================================
    with dot.subgraph(name='cluster_3_logic') as c:
        c.attr(label='4. Business Logic', style='rounded', bgcolor='#E0F7FA')
        c.node('Logic_Calc', 'calculate_progress()', fillcolor='#B2EBF2')
        c.node('Logic_Notify', 'add_notification()', fillcolor='#B2EBF2')
        c.node('Logic_Save', 'save_data()', fillcolor='#B2EBF2')
        c.node('Logic_SkillMatch', 'get_skill_match()', fillcolor='#B2EBF2') # Added Skill Match logic
        c.node('Logic_Generate', 'generate_sample_data()', fillcolor='#B2EBF2') # Added data generation

    # ======================================================
    # E. AI MODELS (Fifth Level)
    # ======================================================
    with dot.subgraph(name='cluster_4_models') as c:
        c.attr(label='5. AI Models', style='rounded', bgcolor='#F3E5F5')
        c.node('Model_Skill', 'Skill Extraction', fillcolor='#E1BEE7')
        c.node('Model_Rec', 'Recommendation Engine', fillcolor='#E1BEE7')
        c.node('Model_Pred', 'Resource Prediction', fillcolor='#E1BEE7')

    # ======================================================
    # F. DATA STORAGE (Bottom Level - Restored all nodes)
    # ======================================================
    with dot.subgraph(name='cluster_5_data') as c:
        c.attr(label='6. Data Storage (Session State)', style='rounded,dashed', bgcolor='#F5F5F5')

        # Primary Data
        c.node('Data_Emp', 'employees_data', shape='cylinder', fillcolor='lightgray')
        c.node('Data_Proj', 'projects_data', shape='cylinder', fillcolor='lightgray')
        c.node('Data_Task', 'tasks_data', shape='cylinder', fillcolor='lightgray')

        # Auxiliary Data (Restored)
        c.node('Data_Notify', 'notifications_data', shape='cylinder', fillcolor='lightgray')
        c.node('Data_Doc', 'documents_data', shape='cylinder', fillcolor='lightgray')
        c.node('Data_Issue', 'issues_data', shape='cylinder', fillcolor='lightgray')

        # UI State (Restored)
        c.node('Data_UI_State', 'UI State Management', shape='note', fillcolor='thistle')

    # ======================================================
    # LOGICAL CONNECTIONS
    # ======================================================

    # UI Interaction
    dot.edge('PMS_Run', 'Dashboard_UI', label='default')
    dot.edge('Proj_UI', 'Proj_Select')
    dot.edge('Task_UI', 'Task_Select')
    dot.edge('Docs_UI', 'Data_Doc', label='C/R/U', style='dotted', color='forestgreen')
    dot.edge('Docs_UI', 'Data_Issue', label='C/R/U', style='dotted', color='forestgreen')
    dot.edge('Docs_UI', 'Logic_Notify', color='firebrick', style='bold', label='issue events')
    dot.edge('Sidebar_UI', 'Data_Notify', label='reads', style='dotted', color='forestgreen')
    dot.edge('Proj_UI', 'Data_UI_State', label='updates', style='dotted', color='darkorchid')

    # Workflow to Models
    dot.edge('Proj_Rec', 'Model_Rec', color='blue')
    dot.edge('Proj_Rec', 'Model_Skill', color='blue')
    dot.edge('Task_Predict', 'Model_Pred', color='blue')

    # Workflow/Analytics to Logic
    dot.edge('Proj_Assign', 'Logic_Save')
    dot.edge('Task_Assign', 'Logic_Save')
    dot.edge('Analytics_UI', 'Logic_Calc', label='metrics')
    dot.edge('Analytics_UI', 'Model_Pred', label='request') # Analytics uses Prediction Model

    # Logic & Data Persistence
    dot.edge('Logic_Save', 'Data_Proj', style='dotted', color='forestgreen', label='writes')
    dot.edge('Logic_Save', 'Data_Task', style='dotted', color='forestgreen', label='writes')
    dot.edge('Logic_Save', 'Logic_Notify')
    dot.edge('Logic_Notify', 'Data_Notify', style='dotted', color='forestgreen', label='writes')
    dot.edge('Logic_Calc', 'Data_Task', style='dotted', color='forestgreen', label='reads')
    dot.edge('Logic_SkillMatch', 'Data_Emp', style='dotted', color='forestgreen', label='reads')

    # ======================================================
    # VERTICAL SPINE (The "Secret Sauce" for Vertical Layout)
    # ======================================================
    # These invisible edges force the clusters to stack Top-to-Bottom

    # Core -> UI
    dot.edge('PMS_Run', 'Dashboard_UI', style='invis', weight='10')
    # UI -> Workflows (Connect a UI node to a workflow node)
    dot.edge('Sidebar_UI', 'Proj_Select', style='invis', weight='10')
    # Workflows -> Logic
    dot.edge('Task_Assign', 'Logic_Calc', style='invis', weight='10')
    # Logic -> Models
    dot.edge('Logic_Generate', 'Model_Skill', style='invis', weight='10')
    # Models -> Data
    dot.edge('Model_Pred', 'Data_Proj', style='invis', weight='10')

    # Render
    output = dot.render('vertical_system_diagram', view=False, cleanup=True)
    print(f"Vertical diagram generated successfully: {output}")
    return output

if __name__ == "__main__":
    create_vertical_system_diagram()
```

```python
from graphviz import Digraph

def create_vertical_system_diagram():
    """
    Generates a vertical, stacked System Function Call Diagram, restoring all original nodes.
    Target aspect ratio: Tall and Narrow.
    """
    dot = Digraph(
        comment='Project Management System - Vertical Stack',
        format='png'
    )

    # 1. Layout Settings for Verticality
    dot.attr(
        rankdir='TB',        # Top to Bottom (Vertical Layout)
        splines='ortho',     # Orthogonal lines (cleaner corners)
        nodesep='0.6',       # Space between nodes
        ranksep='0.8',       # Space between layers
        concentrate='true',  # Merge lines to reduce clutter
        dpi='300'            # High resolution
    )

    # 2. Node Styles
    dot.attr('node', shape='box', style='rounded,filled', fontname='Arial', fontsize='10', margin='0.2')
    dot.attr('edge', fontname='Arial', fontsize='9')

    # ======================================================
    # A. CORE SYSTEM (Top Level)
    # ======================================================
    with dot.subgraph(name='cluster_0_core') as c:
        c.attr(label='1. Core System', style='rounded', bgcolor='#E3F2FD')
        c.node('PMS_Init', 'ProjectManagementSystem.__init__()', fillcolor='#2196F3', fontcolor='white')
        c.node('PMS_Run', 'ProjectManagementSystem.run()', fillcolor='#2196F3', fontcolor='white')
        c.edge('PMS_Init', 'PMS_Run', label='starts')

    # ======================================================
    # B. UI LAYER (Second Level - All main render functions)
    # ======================================================
    with dot.subgraph(name='cluster_1_ui') as c:
        c.attr(label='2. User Interface', style='rounded', bgcolor='#E1F5FE')

        c.node('Dashboard_UI', 'render_dashboard()', fillcolor='#B3E5FC')
        c.node('Emp_UI', 'render_employee_management()', fillcolor='#B3E5FC')
        c.node('Proj_UI', 'render_project_workflow()', fillcolor='#B3E5FC')
        c.node('Task_UI', 'render_task_workflow()', fillcolor='#B3E5FC')
        c.node('Analytics_UI', 'render_analytics_reports()', fillcolor='#B3E5FC') # Corrected name
        c.node('Docs_UI', 'render_documents_issues()', fillcolor='#B3E5FC')     # Re-added
        c.node('Sidebar_UI', 'render_sidebar()', fillcolor='#B3E5FC')           # Added sidebar render

    # ======================================================
    # C. WORKFLOWS (Third Level)
    # ======================================================
    with dot.subgraph(name='cluster_2_workflows') as c:
        c.attr(label='3. Workflows', style='rounded', bgcolor='#FFFDE7')

        # Project Workflow
        c.node('Proj_Select', 'Proj: Select Action', fillcolor='#FFF9C4')
        c.node('Proj_Edit', 'Proj: Edit Details', fillcolor='#FFF9C4')
        c.node('Proj_Rec', 'Proj: AI Recommendation', fillcolor='#FFF9C4')
        c.node('Proj_Assign', 'Proj: Assignment', fillcolor='#FFF9C4')
        c.edge('Proj_Select', 'Proj_Edit')
        c.edge('Proj_Edit', 'Proj_Rec')
        c.edge('Proj_Rec', 'Proj_Assign')

        # Task Workflow
        c.node('Task_Select', 'Task: Select Action', fillcolor='#FFF9C4')
        c.node('Task_Details', 'Task: Edit Details', fillcolor='#FFF9C4')
        c.node('Task_Predict', 'Task: AI Prediction', fillcolor='#FFF9C4')
        c.node('Task_Assign', 'Task: Assignment', fillcolor='#FFF9C4')
        c.edge('Task_Select', 'Task_Details')
        c.edge('Task_Details', 'Task_Predict')
        c.edge('Task_Predict', 'Task_Assign')

    # ======================================================
    # D. BUSINESS LOGIC (Fourth Level)
    # ======================================================
    with dot.subgraph(name='cluster_3_logic') as c:
        c.attr(label='4. Business Logic', style='rounded', bgcolor='#E0F7FA')
        c.node('Logic_Calc', 'calculate_progress()', fillcolor='#B2EBF2')
        c.node('Logic_Notify', 'add_notification()', fillcolor='#B2EBF2')
        c.node('Logic_Save', 'save_data()', fillcolor='#B2EBF2')
        c.node('Logic_SkillMatch', 'get_skill_match()', fillcolor='#B2EBF2') # Added Skill Match logic
        c.node('Logic_Generate', 'generate_sample_data()', fillcolor='#B2EBF2') # Added data generation

    # ======================================================
    # E. AI MODELS (Fifth Level)
    # ======================================================
    with dot.subgraph(name='cluster_4_models') as c:
        c.attr(label='5. AI Models', style='rounded', bgcolor='#F3E5F5')
        c.node('Model_Skill', 'Skill Extraction', fillcolor='#E1BEE7')
        c.node('Model_Rec', 'Recommendation Engine', fillcolor='#E1BEE7')
        c.node('Model_Pred', 'Resource Prediction', fillcolor='#E1BEE7')

    # ======================================================
    # F. DATA STORAGE (Bottom Level - Restored all nodes)
    # ======================================================
    with dot.subgraph(name='cluster_5_data') as c:
        c.attr(label='6. Data Storage (Session State)', style='rounded,dashed', bgcolor='#F5F5F5')

        # Primary Data
        c.node('Data_Emp', 'employees_data', shape='cylinder', fillcolor='lightgray')
        c.node('Data_Proj', 'projects_data', shape='cylinder', fillcolor='lightgray')
        c.node('Data_Task', 'tasks_data', shape='cylinder', fillcolor='lightgray')

        # Auxiliary Data (Restored)
        c.node('Data_Notify', 'notifications_data', shape='cylinder', fillcolor='lightgray')
        c.node('Data_Doc', 'documents_data', shape='cylinder', fillcolor='lightgray')
        c.node('Data_Issue', 'issues_data', shape='cylinder', fillcolor='lightgray')

        # UI State (Restored)
        c.node('Data_UI_State', 'UI State Management', shape='note', fillcolor='thistle')

    # ======================================================
    # LOGICAL CONNECTIONS
    # ======================================================

    # UI Interaction
    dot.edge('PMS_Run', 'Dashboard_UI', label='default')
    dot.edge('Proj_UI', 'Proj_Select')
    dot.edge('Task_UI', 'Task_Select')
    dot.edge('Docs_UI', 'Data_Doc', label='C/R/U', style='dotted', color='forestgreen')
    dot.edge('Docs_UI', 'Data_Issue', label='C/R/U', style='dotted', color='forestgreen')
    dot.edge('Docs_UI', 'Logic_Notify', color='firebrick', style='bold', label='issue events')
    dot.edge('Sidebar_UI', 'Data_Notify', label='reads', style='dotted', color='forestgreen')
    dot.edge('Proj_UI', 'Data_UI_State', label='updates', style='dotted', color='darkorchid')

    # Workflow to Models
    dot.edge('Proj_Rec', 'Model_Rec', color='blue')
    dot.edge('Proj_Rec', 'Model_Skill', color='blue')
    dot.edge('Task_Predict', 'Model_Pred', color='blue')

    # Workflow/Analytics to Logic
    dot.edge('Proj_Assign', 'Logic_Save')
    dot.edge('Task_Assign', 'Logic_Save')
    dot.edge('Analytics_UI', 'Logic_Calc', label='metrics')
    dot.edge('Analytics_UI', 'Model_Pred', label='request') # Analytics uses Prediction Model

    # Logic & Data Persistence
    dot.edge('Logic_Save', 'Data_Proj', style='dotted', color='forestgreen', label='writes')
    dot.edge('Logic_Save', 'Data_Task', style='dotted', color='forestgreen', label='writes')
    dot.edge('Logic_Save', 'Logic_Notify')
    dot.edge('Logic_Notify', 'Data_Notify', style='dotted', color='forestgreen', label='writes')
    dot.edge('Logic_Calc', 'Data_Task', style='dotted', color='forestgreen', label='reads')
    dot.edge('Logic_SkillMatch', 'Data_Emp', style='dotted', color='forestgreen', label='reads')

    # ======================================================
    # VERTICAL SPINE (The "Secret Sauce" for Vertical Layout)
    # ======================================================
    # These invisible edges force the clusters to stack Top-to-Bottom

    # Core -> UI
    dot.edge('PMS_Run', 'Dashboard_UI', style='invis', weight='10')
    # UI -> Workflows (Connect a UI node to a workflow node)
    dot.edge('Sidebar_UI', 'Proj_Select', style='invis', weight='10')
    # Workflows -> Logic
    dot.edge('Task_Assign', 'Logic_Calc', style='invis', weight='10')
    # Logic -> Models
    dot.edge('Logic_Generate', 'Model_Skill', style='invis', weight='10')
    # Models -> Data
    dot.edge('Model_Pred', 'Data_Proj', style='invis', weight='10')

    # Render
    output = dot.render('vertical_system_diagram', view=False, cleanup=True)
    print(f"Vertical diagram generated successfully: {output}")
    return output

if __name__ == "__main__":
    create_vertical_system_diagram()
```
