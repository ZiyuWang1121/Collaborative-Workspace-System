```python
from graphviz import Digraph

def create_project_system_db_diagram():
    """
    Generates a Database Entity-Relationship (ER) Diagram
    based on the data structures visualized in the uploaded schema image.
    The layout uses HTML tables within graphviz nodes for a rich visual representation.
    """
    # Initialize Digraph
    dot = Digraph(comment='Collaborative Workspace Data Schema', format='png')
    # Set overall graph attributes
    dot.attr(rankdir='TB', size='14,10', dpi='300', nodesep='0.6', ranksep='0.8')
    dot.attr('node', fontname='Arial', fontsize='12', shape='plaintext')

    # --- Color Palette (Approximation based on image) ---
    COLOR_EMPLOYEES = '#E1F5FE'  # Light Blue
    COLOR_PROJECTS = '#FFF3E0'   # Light Orange
    COLOR_TASKS = '#F3E5F5'      # Light Purple
    COLOR_DOCUMENTS = '#F1F8E9'  # Light Green
    COLOR_ISSUES = '#FFEBEE'     # Light Red/Pink
    COLOR_NOTIFICATIONS = '#FFFDE7' # Light Yellow/Beige

    # --- Node Definitions (Tables) ---

    # 1. Employees Table
    dot.node('Employees', f'''<
    <table border="0" cellborder="1" cellspacing="0" cellpadding="4">
        <tr><td bgcolor="{COLOR_EMPLOYEES}" colspan="2"><b><font point-size="14">EMPLOYEES</font></b></td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>PK</b> employee_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">name</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">job_title</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">department</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">email</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">skills</td><td align="left" bgcolor="#ffffff">TEXT</td></tr>
        <tr><td align="left" bgcolor="#ffffff">performance_rating</td><td align="left" bgcolor="#ffffff">FLOAT</td></tr>
        <tr><td align="left" bgcolor="#ffffff">experience_years</td><td align="left" bgcolor="#ffffff">INT</td></tr>
    </table>
    >''')

    # 2. Projects Table
    dot.node('Projects', f'''<
    <table border="0" cellborder="1" cellspacing="0" cellpadding="4">
        <tr><td bgcolor="{COLOR_PROJECTS}" colspan="2"><b><font point-size="14">PROJECTS</font></b></td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>PK</b> project_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>FK</b> manager_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">project_name</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">description</td><td align="left" bgcolor="#ffffff">TEXT</td></tr>
        <tr><td align="left" bgcolor="#ffffff">status</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">start_date</td><td align="left" bgcolor="#ffffff">DATE</td></tr>
        <tr><td align="left" bgcolor="#ffffff">deadline</td><td align="left" bgcolor="#ffffff">DATE</td></tr>
        <tr><td align="left" bgcolor="#ffffff">complexity_score</td><td align="left" bgcolor="#ffffff">INT</td></tr>
        <tr><td align="left" bgcolor="#ffffff">priority</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
    </table>
    >''')

    # 3. Tasks Table
    dot.node('Tasks', f'''<
    <table border="0" cellborder="1" cellspacing="0" cellpadding="4">
        <tr><td bgcolor="{COLOR_TASKS}" colspan="2"><b><font point-size="14">TASKS</font></b></td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>PK</b> task_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>FK</b> project_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>FK</b> assigned_to_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">task_name</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">task_description</td><td align="left" bgcolor="#ffffff">TEXT</td></tr>
        <tr><td align="left" bgcolor="#ffffff">status</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">complexity</td><td align="left" bgcolor="#ffffff">INT</td></tr>
        <tr><td align="left" bgcolor="#ffffff">estimated_duration</td><td align="left" bgcolor="#ffffff">INT</td></tr>
    </table>
    >''')

    # 4. Documents Table
    dot.node('Documents', f'''<
    <table border="0" cellborder="1" cellspacing="0" cellpadding="4">
        <tr><td bgcolor="{COLOR_DOCUMENTS}" colspan="2"><b><font point-size="14">DOCUMENTS</font></b></td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>PK</b> doc_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>FK</b> project_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">title</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">file_name</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">file_type</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">uploaded_at</td><td align="left" bgcolor="#ffffff">TIMESTAMP</td></tr>
    </table>
    >''')

    # 5. Issues Table
    dot.node('Issues', f'''<
    <table border="0" cellborder="1" cellspacing="0" cellpadding="4">
        <tr><td bgcolor="{COLOR_ISSUES}" colspan="2"><b><font point-size="14">ISSUES</font></b></td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>PK</b> issue_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>FK</b> project_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">title</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">priority</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">status</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">reported_at</td><td align="left" bgcolor="#ffffff">TIMESTAMP</td></tr>
    </table>
    >''')

    # 6. Notifications Table (New based on the image)
    dot.node('Notifications', f'''<
    <table border="0" cellborder="1" cellspacing="0" cellpadding="4">
        <tr><td bgcolor="{COLOR_NOTIFICATIONS}" colspan="2"><b><font point-size="14">NOTIFICATIONS</font></b></td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>PK</b> notification_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>FK</b> project_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff"><b>FK</b> task_id</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">message</td><td align="left" bgcolor="#ffffff">TEXT</td></tr>
        <tr><td align="left" bgcolor="#ffffff">type</td><td align="left" bgcolor="#ffffff">VARCHAR</td></tr>
        <tr><td align="left" bgcolor="#ffffff">timestamp</td><td align="left" bgcolor="#ffffff">TIMESTAMP</td></tr>
        <tr><td align="left" bgcolor="#ffffff">read</td><td align="left" bgcolor="#ffffff">BOOLEAN</td></tr>
    </table>
    >''')

    # 7. Relationship Legend Node (Added to match the image)
    dot.node('Legend', f'''<
    <table border="0" cellborder="1" cellspacing="0" cellpadding="4">
        <tr><td bgcolor="#F5F5F5" colspan="2"><b><font point-size="14">RELATIONSHIP LEGEND</font></b></td></tr>

        <tr><td align="left" bgcolor="#ffffff">Management and Team</td><td align="left" bgcolor="#ffffff"><font color="#4FC3F7">Teal</font></td></tr>
        <tr><td align="left" bgcolor="#ffffff">Project Artifacts</td><td align="left" bgcolor="#ffffff"><font color="#8BC34A">Green</font></td></tr>
        <tr><td align="left" bgcolor="#ffffff">Task Relationships</td><td align="left" bgcolor="#ffffff"><font color="#9C27B0">Purple</font></td></tr>
        <tr><td align="left" bgcolor="#ffffff">Issue Tracking</td><td align="left" bgcolor="#ffffff"><font color="#F44336">Red</font></td></tr>
        <tr><td align="left" bgcolor="#ffffff">Notifications</td><td align="left" bgcolor="#ffffff"><font color="#FF9800">Orange</font></td></tr>

        <tr><td align="left" bgcolor="#cccccc" colspan="2"><b>CARDINALITY (In Label)</b></td></tr>
        <tr><td align="left" bgcolor="#ffffff">One-to-One (1:1)</td><td align="left" bgcolor="#ffffff">Solid Line + Label</td></tr>
        <tr><td align="left" bgcolor="#ffffff">One-to-Many (1:N)</td><td align="left" bgcolor="#ffffff">Solid Line + Label</td></tr>
        <tr><td align="left" bgcolor="#ffffff">Many-to-Many (M:N)</td><td align="left" bgcolor="#ffffff">Solid Line + Label</td></tr>
    </table>
    >''')


    # --- Relationships (Edges) ---

    # 1. Management and Team Relationships (Blue/Teal)
    # Employee manages/is on the team for Project (via FK manager_id)
    dot.edge('Employees', 'Projects', label='Manages / Team\n(1:1) / (M:N)', color='#4FC3F7', fontcolor='#4FC3F7')

    # Employee assigned to Task (via FK assigned_to_id)
    dot.edge('Employees', 'Tasks', label='Assigned To\n(N:1)', color='#9C27B0', fontcolor='#9C27B0')

    # 2. Project Artifacts Relationships (Red, Green, Purple)
    # Project has Documents
    dot.edge('Projects', 'Documents', label='Has\n(1:N)', color='#8BC34A', fontcolor='#8BC34A')

    # Project has Issues
    dot.edge('Projects', 'Issues', label='Has\n(1:N)', color='#F44336', fontcolor='#F44336')

    # Project contains Tasks
    dot.edge('Projects', 'Tasks', label='Contains\n(1:N)', color='#9C27B0', fontcolor='#9C27B0')


    # 3. Notification Flow (Orange) - Note: Arrows originate from the Artifacts, pointing to the Notifications table
    # Issues generates Notifications
    dot.edge('Issues', 'Notifications', label='Generates\n(1:N)', color='#FF9800', fontcolor='#FF9800')

    # Documents generates Notifications
    dot.edge('Documents', 'Notifications', label='Generates\n(1:N)', color='#FF9800', fontcolor='#FF9800')

    # Tasks generates Notifications
    dot.edge('Tasks', 'Notifications', label='Generates\n(1:N)', color='#FF9800', fontcolor='#FF9800')

    # 4. Projects generates Notifications (via FK project_id)
    # This arrow is shown coming from Projects to Notifications in the image.
    dot.edge('Projects', 'Notifications', label='Generates\n(1:N)', color='#FF9800', fontcolor='#FF9800')


    # Render the graph
    # In a real environment, you would call:
    # output_file = dot.render('project_system_schema', view=False, cleanup=True)
    # print(f"Schema Diagram generated: {output_file}")
    return dot

if __name__ == '__main__':
    # This block allows you to run the script locally to generate the image file.
    # When running on the server, you would typically handle the output differently.
    try:
        diagram = create_project_system_db_diagram()
        diagram.render('project_system_schema', view=False, cleanup=True)
        print("Diagram generated as project_system_schema.png (or .gv file).")
    except ImportError:
        print("Please install graphviz to run this script: pip install graphviz")
```

```python
from graphviz import Digraph

def create_optimized_system_function_call_diagram():
    """
    Optimized System Function Call Diagram - Better layout and organization
    """
    dot = Digraph(
        comment='Project Management System - Optimized Function Call Flow',
        format='png'
    )

    dot.attr(
        rankdir='TB',
        size='24,32',  # Increase canvas size
        dpi='300',
        nodesep='0.8',  # Increase node spacing
        ranksep='1.5',  # Increase rank spacing
        splines='ortho',
        concentrate='false'  # Prevent line merging
    )

    dot.attr('node',
            shape='box',
            style='rounded,filled',
            fontname='Arial',
            fontsize='10',
            margin='0.3')

    dot.attr('edge',
            fontname='Arial',
            fontsize='9',
            labeldistance='1.5',
            labelfloat='false',
            labelloc='c')

    # ======================================================
    # 1. Core System Initialization (Top)
    # ======================================================
    with dot.subgraph(name='cluster_core') as c:
        c.attr(label='Core System Initialization', style='rounded', bgcolor='lightblue1')
        c.attr(rank='min')  # Force to top

        c.node('PMS_Init', 'ProjectManagementSystem.__init__()', fillcolor='dodgerblue', fontcolor='white')
        c.node('PMS_Run', 'ProjectManagementSystem.run()', fillcolor='dodgerblue', fontcolor='white')

        # Core system connections
        c.edge('PMS_Init', 'PMS_Run', label='starts', labeldistance='1.2')

    # ======================================================
    # 2. Main UI Interface Layer (Second Layer)
    # ======================================================
    with dot.subgraph(name='cluster_ui') as c:
        c.attr(label='Main User Interface', style='rounded', bgcolor='lightsteelblue1')
        c.attr(rank='same')

        c.node('Dashboard_UI', 'render_dashboard()', fillcolor='lightsteelblue1')
        c.node('Emp_UI', 'render_employee_management()', fillcolor='lightsteelblue1')
        c.node('Proj_UI', 'render_project_workflow()', fillcolor='lightsteelblue1')
        c.node('Task_UI', 'render_task_workflow()', fillcolor='lightsteelblue1')
        c.node('Analytics_UI', 'render_analytics_reports()', fillcolor='lightsteelblue1')
        c.node('Docs_UI', 'render_documents_issues()', fillcolor='lightsteelblue1')

    # ======================================================
    # 3. Project Workflow Layer (Left Vertical Flow)
    # ======================================================
    with dot.subgraph(name='cluster_project_workflow') as c:
        c.attr(label='Project Workflow Steps', style='rounded', bgcolor='lightyellow')

        # Project workflow steps - vertical arrangement
        c.node('Proj_Select_Action', '1. Select Action\n_render_project_select_action_step()', fillcolor='lightsteelblue2')
        c.node('Proj_Edit', '2. Project Details\n_render_edit_project_step()', fillcolor='lightsteelblue2')
        c.node('Proj_Recommend', '3. AI Recommendations\n_render_recommendations_step()', fillcolor='lightsteelblue2')
        c.node('Proj_Assign', '4. Team Assignment\n_render_assignment_step()', fillcolor='lightsteelblue2')
        c.node('Proj_Complete', '5. Completion\n_render_complete_step()', fillcolor='lightsteelblue2')

        # Project workflow sequence
        c.edge('Proj_Select_Action', 'Proj_Edit', label='new/edit project', labeldistance='1.3')
        c.edge('Proj_Edit', 'Proj_Recommend', label='continue', labeldistance='1.3')
        c.edge('Proj_Recommend', 'Proj_Assign', label='continue', labeldistance='1.3')
        c.edge('Proj_Assign', 'Proj_Complete', label='confirm', labeldistance='1.3')
        c.edge('Proj_Complete', 'Proj_Select_Action', label='workflow complete', labeldistance='1.3', style='dashed')

    # ======================================================
    # 4. Task Workflow Layer (Right Vertical Flow)
    # ======================================================
    with dot.subgraph(name='cluster_task_workflow') as c:
        c.attr(label='Task Workflow Steps', style='rounded', bgcolor='lightyellow')

        # Task workflow steps - vertical arrangement
        c.node('Task_Select_Action', '1. Select Action\n_render_task_select_action_step()', fillcolor='lightsteelblue2')
        c.node('Task_Details', '2. Task Details\n_render_task_details_step()', fillcolor='lightsteelblue2')
        c.node('Task_Predict', '3. AI Predictions\n_render_task_predictions_step()', fillcolor='lightsteelblue2')
        c.node('Task_Assign', '4. Assignment\n_render_task_assignment_step()', fillcolor='lightsteelblue2')
        c.node('Task_Complete', '5. Completion\n_render_task_complete_step()', fillcolor='lightsteelblue2')

        # Task workflow sequence
        c.edge('Task_Select_Action', 'Task_Details', label='new/edit task', labeldistance='1.3')
        c.edge('Task_Details', 'Task_Predict', label='continue', labeldistance='1.3')
        c.edge('Task_Predict', 'Task_Assign', label='continue', labeldistance='1.3')
        c.edge('Task_Assign', 'Task_Complete', label='confirm', labeldistance='1.3')
        c.edge('Task_Complete', 'Task_Select_Action', label='workflow complete', labeldistance='1.3', style='dashed')

    # ======================================================
    # 5. Core Business Logic (Middle Layer)
    # ======================================================
    with dot.subgraph(name='cluster_business_logic') as c:
        c.attr(label='Core Business Logic', style='rounded', bgcolor='lightskyblue1')

        # Progress calculation and skill matching
        c.node('Calc_Progress', 'calculate_project_progress()', fillcolor='lightskyblue1')
        c.node('Skill_Match', 'get_skill_match_recommendations()', fillcolor='lightskyblue1')
        c.node('Add_Notify', 'add_notification()', fillcolor='lightskyblue1')
        c.node('Gen_Sample', 'generate_sample_data()', fillcolor='lightskyblue1')

        # Team assignment logic
        c.node('Update_Team', 'update_team_assignment()', fillcolor='lightsalmon')
        c.node('Multi_Update', '_update_from_multiselect()', fillcolor='lightsalmon')
        c.node('Get_Emp_Map', 'get_employee_name_map()', fillcolor='lightsalmon')
        c.node('Get_Team_Members', 'get_project_team_members()', fillcolor='lightsalmon')

        # Data persistence
        c.node('Save_Proj_Data', '_save_project_data()', fillcolor='lightgreen')
        c.node('Save_Task_Data', '_save_task_data()', fillcolor='lightgreen')

        # Workflow management
        c.node('Cancel_Proj', '_cancel_project_workflow()', fillcolor='lightcoral')
        c.node('Cancel_Task', '_cancel_task_workflow()', fillcolor='lightcoral')

    # ======================================================
    # 6. External AI Models (Top right)
    # ======================================================
    with dot.subgraph(name='cluster_models') as c:
        c.attr(label='External AI / ML Models', style='rounded', bgcolor='bisque')
        c.attr(rank='max')

        c.node('Skills_Model', 'Skills Extraction Model', fillcolor='sandybrown')
        c.node('Rec_Model', 'Team Recommendation Model', fillcolor='sandybrown')
        c.node('Task_Model', 'Task Estimation Model', fillcolor='sandybrown')
        c.node('Pred_Model', 'General Prediction Model', fillcolor='sandybrown')

    # ======================================================
    # 7. External Model Interfaces (Connection layer)
    # ======================================================
    with dot.subgraph(name='cluster_model_interfaces') as c:
        c.attr(label='External Model Interfaces', style='rounded', bgcolor='deepskyblue1')

        c.node('Get_Skills', 'get_skills_from_external_model()', fillcolor='deepskyblue1')
        c.node('Get_Recs', 'get_recommendations()', fillcolor='deepskyblue1')
        c.node('Get_Task_Recs', 'get_task_recommendations()', fillcolor='deepskyblue1')
        c.node('Get_Pred', 'get_predictions()', fillcolor='deepskyblue1')

    # ======================================================
    # 8. Data Storage Layer (Bottom)
    # ======================================================
    with dot.subgraph(name='cluster_data') as c:
        c.attr(label='Data Storage (Session State)', style='rounded,dashed', bgcolor='whitesmoke')
        c.attr(rank='sink')  # Force to bottom

        # Primary data storage
        c.node('Emp_Data', 'employees_data', shape='cylinder', fillcolor='lightgray')
        c.node('Proj_Data', 'projects_data', shape='cylinder', fillcolor='lightgray')
        c.node('Task_Data', 'tasks_data', shape='cylinder', fillcolor='lightgray')
        c.node('Notify_Data', 'notifications_data', shape='cylinder', fillcolor='lightgray')

        # Document and Issue data
        c.node('Doc_Data', 'documents_data', shape='cylinder', fillcolor='lightgray')
        c.node('Issue_Data', 'issues_data', shape='cylinder', fillcolor='lightgray')

        # UI state management
        c.node('UI_State', 'UI State Management\nworkflow_step, form_data', shape='note', fillcolor='thistle')

    # ======================================================
    # 9. External Libraries (Very bottom)
    # ======================================================
    with dot.subgraph(name='cluster_libs') as c:
        c.attr(label='External Libraries & Dependencies', style='rounded,dotted', bgcolor='honeydew')
        c.attr(rank='max')

        c.node('Streamlit', 'Streamlit Framework', fillcolor='palegreen')
        c.node('Pandas', 'Pandas Library', fillcolor='palegreen')
        c.node('Plotly', 'Plotly Visualization', fillcolor='palegreen')
        c.node('DateTime', 'Date/Time Handling', fillcolor='palegreen')
        c.node('UUID', 'UUID Generation', fillcolor='palegreen')

    # ======================================================
    # Cross-cluster connections - reorganized
    # ======================================================

    # 1. Core System to UI Layer
    dot.edge('PMS_Run', 'Dashboard_UI', label='default tab', labeldistance='1.2')
    dot.edge('PMS_Run', 'Emp_UI', label='tab navigation', labeldistance='1.2')
    dot.edge('PMS_Run', 'Proj_UI', label='tab navigation', labeldistance='1.2')
    dot.edge('PMS_Run', 'Task_UI', label='tab navigation', labeldistance='1.2')
    dot.edge('PMS_Run', 'Analytics_UI', label='tab navigation', labeldistance='1.2')
    dot.edge('PMS_Run', 'Docs_UI', label='tab navigation', labeldistance='1.2')

    # 2. UI to Workflow Entry
    dot.edge('Proj_UI', 'Proj_Select_Action', label='workflow start', labeldistance='1.3')
    dot.edge('Task_UI', 'Task_Select_Action', label='workflow start', labeldistance='1.3')

    # 3. Workflow to Business Logic
    dot.edge('Proj_Recommend', 'Get_Recs', label='team recommendations', labeldistance='1.4')
    dot.edge('Proj_Recommend', 'Get_Skills', label='skill extraction', labeldistance='1.4')
    dot.edge('Task_Predict', 'Get_Task_Recs', label='duration/budget', labeldistance='1.4')

    dot.edge('Proj_Recommend', 'Update_Team', label='AI recommendations', labeldistance='1.4')
    dot.edge('Update_Team', 'Multi_Update', label='sync state', labeldistance='1.4')
    dot.edge('Multi_Update', 'Get_Emp_Map', label='name lookup', labeldistance='1.4')
    dot.edge('Task_Predict', 'Get_Team_Members', label='project team', labeldistance='1.4')

    # 4. Workflow to Data Persistence
    dot.edge('Proj_Assign', 'Save_Proj_Data', label='save project', labeldistance='1.4')
    dot.edge('Task_Assign', 'Save_Task_Data', label='save task', labeldistance='1.4')

    # 5. Model Interface to External Model
    dot.edge('Get_Skills', 'Skills_Model', label='calls', color='mediumblue', penwidth='2.5', labeldistance='1.6')
    dot.edge('Get_Recs', 'Rec_Model', label='calls', color='mediumblue', penwidth='2.5', labeldistance='1.6')
    dot.edge('Get_Task_Recs', 'Task_Model', label='calls', color='mediumblue', penwidth='2.5', labeldistance='1.6')
    dot.edge('Get_Pred', 'Pred_Model', label='calls', color='mediumblue', penwidth='2.5', labeldistance='1.6')

    # 6. Business Logic to Data Analysis
    dot.edge('Analytics_UI', 'Calc_Progress', label='metrics calculation', labeldistance='1.4')
    dot.edge('Analytics_UI', 'Get_Pred', label='predictions', labeldistance='1.4')
    dot.edge('Dashboard_UI', 'Calc_Progress', label='progress display', labeldistance='1.4')

    # 7. Data Access Pattern
    dot.edge('Emp_UI', 'Emp_Data', label='C/R/U', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Proj_UI', 'Proj_Data', label='C/R/U', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Task_UI', 'Task_Data', label='C/R/U', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Docs_UI', 'Doc_Data', label='C/R', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Docs_UI', 'Issue_Data', label='C/R/U', style='dotted', color='forestgreen', labeldistance='1.4')

    dot.edge('Calc_Progress', 'Task_Data', label='reads', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Skill_Match', 'Emp_Data', label='reads', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Get_Emp_Map', 'Emp_Data', label='reads', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Get_Team_Members', 'Proj_Data', label='reads', style='dotted', color='forestgreen', labeldistance='1.4')

    dot.edge('Save_Proj_Data', 'Proj_Data', label='writes', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Save_Task_Data', 'Task_Data', label='writes', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Gen_Sample', 'Emp_Data', label='writes', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Gen_Sample', 'Proj_Data', label='writes', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Gen_Sample', 'Task_Data', label='writes', style='dotted', color='forestgreen', labeldistance='1.4')
    dot.edge('Gen_Sample', 'Notify_Data', label='writes', style='dotted', color='forestgreen', labeldistance='1.4')

    # 8. Notification Generation
    dot.edge('Emp_UI', 'Add_Notify', label='employee updates', color='firebrick', style='bold', penwidth='2.5', labeldistance='1.5')
    dot.edge('Proj_UI', 'Add_Notify', label='project events', color='firebrick', style='bold', penwidth='2.5', labeldistance='1.5')
    dot.edge('Task_UI', 'Add_Notify', label='task events', color='firebrick', style='bold', penwidth='2.5', labeldistance='1.5')
    dot.edge('Docs_UI', 'Add_Notify', label='doc/issue events', color='firebrick', style='bold', penwidth='2.5', labeldistance='1.5')
    dot.edge('Save_Proj_Data', 'Add_Notify', label='project saved', color='firebrick', style='bold', penwidth='2.5', labeldistance='1.5')
    dot.edge('Save_Task_Data', 'Add_Notify', label='task saved', color='firebrick', style='bold', penwidth='2.5', labeldistance='1.5')
    dot.edge('Add_Notify', 'Notify_Data', label='creates', style='dotted', color='forestgreen', labeldistance='1.4')

    # 9. UI State Management
    dot.edge('PMS_Run', 'UI_State', label='manages', style='dotted', color='darkorchid', labeldistance='1.4')
    dot.edge('Proj_UI', 'UI_State', label='updates', style='dotted', color='darkorchid', labeldistance='1.4')
    dot.edge('Task_UI', 'UI_State', label='updates', style='dotted', color='darkorchid', labeldistance='1.4')
    dot.edge('Cancel_Proj', 'UI_State', label='resets', style='dotted', color='darkorchid', labeldistance='1.4')
    dot.edge('Cancel_Task', 'UI_State', label='resets', style='dotted', color='darkorchid', labeldistance='1.4')

    # 10. Library Dependencies
    dot.edge('PMS_Run', 'Streamlit', label='UI framework', style='dashed', color='darkorange', labeldistance='1.5')
    dot.edge('Analytics_UI', 'Pandas', label='data processing', style='dashed', color='darkorange', labeldistance='1.5')
    dot.edge('Analytics_UI', 'Plotly', label='visualization', style='dashed', color='darkorange', labeldistance='1.5')
    dot.edge('Dashboard_UI', 'Plotly', label='metrics display', style='dashed', color='darkorange', labeldistance='1.5')
    dot.edge('Gen_Sample', 'UUID', label='ID generation', style='dashed', color='darkorange', labeldistance='1.5')
    dot.edge('Calc_Progress', 'DateTime', label='date calculations', style='dashed', color='darkorange', labeldistance='1.5')
    dot.edge('Add_Notify', 'DateTime', label='timestamp', style='dashed', color='darkorange', labeldistance='1.5')

    # ======================================================
    # Legend (Top left)
    # ======================================================
    dot.node(
        'Legend',
        '''<
        <table border="0" cellborder="1" cellspacing="0" cellpadding="4" bgcolor="whitesmoke">
            <tr><td colspan="2" bgcolor="lightgray"><b>Connection Legend</b></td></tr>
            <tr><td align="left">────── Solid Line</td><td align="left">Internal Function Calls</td></tr>
            <tr><td align="left"><font color="mediumblue">━━━━━━</font> Thick Blue</td><td align="left">External Model Integration</td></tr>
            <tr><td align="left"><font color="firebrick">━━━━━━</font> Thick Red</td><td align="left"><b>Notification Generation</b></td></tr>
            <tr><td align="left"><font color="forestgreen">······</font> Dotted Green</td><td align="left">Data Access (Read/Write)</td></tr>
            <tr><td align="left"><font color="darkorchid">······</font> Dotted Purple</td><td align="left">UI State Management</td></tr>
            <tr><td align="left"><font color="darkorange">- - - -</font> Dashed Orange</td><td align="left">Library Dependencies</td></tr>
        </table>
        >''',
        shape='plaintext'
    )

    # Layout constraints
    dot.attr(rank='min')
    dot.edge('Legend', 'PMS_Init', style='invis', minlen='2')

    output = dot.render('system_function_calls', view=False, cleanup=True)
    print(f"Optimized layout diagram generated: {output}")
    return dot

if __name__ == "__main__":
    create_optimized_system_function_call_diagram()
```
